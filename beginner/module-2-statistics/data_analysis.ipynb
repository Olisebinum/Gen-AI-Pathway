# Data Analysis with NumPy & Pandas
## Exploring the Iris Dataset

This notebook demonstrates exploratory data analysis and basic statistics on the classic Iris dataset using Python's `pandas` and `numpy` libraries.


```python
import numpy as np
import pandas as pd
from scipy import stats

# Load the Iris dataset directly from a public URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)

print("Dataset loaded successfully!")
print(f"Shape: {data.shape[0]} rows × {data.shape[1]} columns")
```

---
## 1. Explore the Data


```python
# ── First few rows ──────────────────────────────────────────────────────────
print("First 5 rows of the dataset:")
data.head()
```

```python
# ── Data types and non-null counts ──────────────────────────────────────────
print("Dataset Info:")
data.info()
```

```python
# ── Summary statistics ───────────────────────────────────────────────────────
print("Summary Statistics:")
data.describe()
```

---
## 2. Calculate Basic Statistics


```python
# ── Select only numeric columns ──────────────────────────────────────────────
numeric_cols = data.select_dtypes(include=[np.number])

# Mean
print("=" * 55)
print("MEAN")
print("=" * 55)
print(numeric_cols.mean().to_string())

# Median
print("\n" + "=" * 55)
print("MEDIAN")
print("=" * 55)
print(numeric_cols.median().to_string())

# Mode  (returns a DataFrame; take row 0 for the first mode)
print("\n" + "=" * 55)
print("MODE")
print("=" * 55)
print(numeric_cols.mode().iloc[0].to_string())

# Standard Deviation
print("\n" + "=" * 55)
print("STANDARD DEVIATION")
print("=" * 55)
print(numeric_cols.std().to_string())
```

```python
# ── Combined statistics table ────────────────────────────────────────────────
stats_df = pd.DataFrame({
    "Mean":   numeric_cols.mean(),
    "Median": numeric_cols.median(),
    "Mode":   numeric_cols.mode().iloc[0],
    "Std Dev": numeric_cols.std(),
    "Min":    numeric_cols.min(),
    "Max":    numeric_cols.max(),
    "Range":  numeric_cols.max() - numeric_cols.min(),
})

print("Combined Statistics Table:")
stats_df.round(4)
```

```python
# ── Correlation matrix ───────────────────────────────────────────────────────
print("Pearson Correlation Matrix:")
corr_matrix = numeric_cols.corr()
print(corr_matrix.to_string())
corr_matrix.round(4)
```

```python
# ── Interpret the strongest correlations ────────────────────────────────────
print("Top Correlations (excluding self-correlation):\n")

# Flatten the upper triangle only
corr_pairs = (
    corr_matrix
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)
corr_pairs.columns = ["Feature A", "Feature B", "Correlation"]
corr_pairs["Abs Correlation"] = corr_pairs["Correlation"].abs()
corr_pairs = corr_pairs.sort_values("Abs Correlation", ascending=False).reset_index(drop=True)

print(corr_pairs[["Feature A", "Feature B", "Correlation"]].to_string(index=False))
```

```python
# ── NumPy cross-check ────────────────────────────────────────────────────────
print("NumPy verification for 'sepal_length':")
col = data["sepal_length"].values

print(f"  np.mean   : {np.mean(col):.4f}")
print(f"  np.median : {np.median(col):.4f}")
print(f"  np.std    : {np.std(col, ddof=1):.4f}  (sample std, ddof=1)")
print(f"  np.min    : {np.min(col):.4f}")
print(f"  np.max    : {np.max(col):.4f}")
print(f"  np.var    : {np.var(col, ddof=1):.4f}  (sample variance)")
```

---
## Summary

| Metric | What it tells us |
|--------|-----------------|
| **Mean** | The arithmetic average — sensitive to outliers |
| **Median** | The middle value — robust to outliers |
| **Mode** | The most frequently occurring value |
| **Std Dev** | Average spread of values around the mean |
| **Correlation** | Linear relationship strength between two features (–1 to +1) |

> **Key finding:** `petal_length` and `petal_width` show the strongest positive correlation (~0.96), meaning longer petals tend to be wider petals across the Iris species.
